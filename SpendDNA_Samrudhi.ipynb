```python
def extract_vendor(description):
    description = str(description).upper()

    # P2P transfers
    if ("UPI-AMAN" in description or
        "UPI-ANKIT" in description or
        "UPI-VIKAS" in description or
        "UPI-PRIYA" in description):
        return "P2P Transfer"

    # ATM withdrawals
    if "ATM" in description and "WDL" in description:
        return "Cash Withdrawal"

    # Vendor matching
    for vendor, keywords in vendor_keywords.items():
        for keyword in keywords:
            if keyword in description:
                return vendor

    return "Uncategorised"
```

```python
def get_time_period(hour):

    if 5 <= hour < 12:
        return "Morning"

    elif 12 <= hour < 17:
        return "Afternoon"

    elif 17 <= hour < 21:
        return "Evening"

    else:
        return "Night"

df_clean["TimePeriod"] = df_clean["Hour"].apply(get_time_period)
```

## Feature 8 - Spending Archetype Detection

```python
# Category-wise Spending Percentage

category_total = (
    df_clean[df_clean["Type"] == "debit"]
    .groupby("category")["Amount"]
    .sum()
)

print(category_total)
```

```python
# Spending Archetype

def detect_archetypes(df):
    archetypes = []

    total_debit = df.loc[df["Type"] == "debit", "Amount"].sum()

    # Category spending
    category_spend = (
        df[df["Type"] == "debit"]
        .groupby("category")["Amount"]
        .sum()
    )

    # Vendor spending
    vendor_spend = (
        df[df["Type"] == "debit"]
        .groupby("Vendor")["Amount"]
        .sum()
    )

    # Transaction counts
    category_count = (
        df[df["Type"] == "debit"]
        .groupby("category")
        .size()
    )

    # 1. FOODIE
    food_spend = category_spend.get("Food Delivery", 0)
    food_pct = (food_spend / total_debit * 100) if total_debit else 0

    if food_pct >= 15:
        archetypes.append("Foodie")

    # 2. QUICK COMMERCE JUNKIE
    quick_spend = category_spend.get("Quick Commerce", 0)
    quick_pct = (quick_spend / total_debit * 100) if total_debit else 0

    if quick_pct >= 10:
        archetypes.append("Quick Commerce Junkie")

    # 3. SHOPAHOLIC
    ecommerce_spend = category_spend.get("E-commerce", 0)
    ecommerce_pct = (ecommerce_spend / total_debit * 100) if total_debit else 0

    if ecommerce_pct >= 15:
        archetypes.append("Shopaholic")

    # 4. INVESTOR
    investment_spend = category_spend.get("Investments", 0)
    investment_pct = (investment_spend / total_debit * 100) if total_debit else 0

    if investment_pct >= 10:
        archetypes.append("Investor")

    # 5. LATE-NIGHT SNACKER
    food_df = df[
        (df["Type"] == "debit") &
        (df["category"] == "Food Delivery")
    ].copy()

    if len(food_df) > 0:
        late_night = food_df[
            (food_df["Hour"] >= 21) | (food_df["Hour"] <= 1)
        ]

        late_pct = len(late_night) / len(food_df) * 100

        if late_pct >= 40:
            archetypes.append("Late-Night Snacker")

    # 6. CAB COMMUTER
    transport_spend = category_spend.get("Transport", 0)
    transport_pct = (transport_spend / total_debit * 100) if total_debit else 0

    if transport_pct >= 8:
        archetypes.append("Cab Commuter")

    # 7. SUBSCRIPTION LOVER
    subscription_count = category_count.get("Subscriptions", 0)

    if subscription_count >= 5:
        archetypes.append("Subscription Lover")

    # 8. YOLO SPENDER
    savings_rate = (
        (df.loc[df["Type"] == "credit", "Amount"].sum() - total_debit)
        / df.loc[df["Type"] == "credit", "Amount"].sum() * 100
        if df.loc[df["Type"] == "credit", "Amount"].sum() > 0
        else 0
    )

    if savings_rate < 0:
        archetypes.append("YOLO Spender")

    # 9. DISCIPLINED SAVER
    if savings_rate >= 20:
        archetypes.append("Disciplined Saver")

    return archetypes


archetypes = detect_archetypes(df_clean)

print("\n" + "=" * 65)
print("SPENDING ARCHETYPES")
print("=" * 65)

if archetypes:
    for archetype in archetypes:
        print(f"✓ {archetype}")
else:
    print("No strong archetype detected.")
```

```python
# ================================================================
#                    SpendDNA FINAL REPORT
# ================================================================

name = "RAHUL SHARMA"

start_date = df_clean["Date"].min()
end_date = df_clean["Date"].max()

months = df_clean["Date"].dt.to_period("M").nunique()
transactions = len(df_clean)

print("\n" + "=" * 65)
print(f"              SpendDNA REPORT - {name}")
print("\n"
      f"{months} months - "
      f"{transactions:,} transactions - "
      f"{start_date.strftime('%b')} to {end_date.strftime('%b %Y')}"
)
print("\n" + "=" * 65)


# ================================================================
# EXECUTIVE SUMMARY
# ================================================================

total_income = df_clean.loc[
    df_clean["Type"].str.lower() == "credit", "Amount"
].sum()

total_spending = df_clean.loc[
    df_clean["Type"].str.lower() == "debit", "Amount"
].sum()

net_change = total_income - total_spending

savings_rate = (
    (net_change / total_income) * 100
    if total_income > 0 else 0
)

print("\nEXECUTIVE SUMMARY")
print("-" * 65)

print(f"Total credits    : ₹{total_income:,.0f}")
print(f"Total debits     : ₹{total_spending:,.0f}")

if net_change >= 0:
    print(f"Net change       : ₹{net_change:,.0f} (savings)")
else:
    print(f"Net change       : -₹{abs(net_change):,.0f} (overspending)")

if savings_rate >= 0:
    print(f"Savings rate     : {savings_rate:.1f}%")
else:
    print(f"Savings rate     : {savings_rate:.1f}% (BURNING SAVINGS)")

print(f"Transactions     : {len(df_clean):,}")
print(f"Unique vendors   : {df_clean['Vendor'].nunique():,}")


# ================================================================
# TOP SPENDING CATEGORIES
# ================================================================

category_spend = (
    df_clean[
        df_clean["Type"].str.lower() == "debit"
    ]
    .groupby("category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP CATEGORIES (% of debit total)")
print("-" * 65)

if len(category_spend) > 0:

    max_category = category_spend.max()

    for category, amount in category_spend.head(5).items():

        percentage = (
            amount / total_spending * 100
            if total_spending > 0 else 0
        )

        bar_length = (
            int((amount / max_category) * 20)
            if max_category > 0 else 0
        )

        print(
            f"{category:<22} "
            f"{'#' * bar_length:<20} "
            f"{percentage:>5.1f}% "
            f"₹{amount:>10,.0f}"
        )


# ================================================================
# TOP SPENDING VENDORS
# ================================================================

vendor_data = df_clean[
    df_clean["Type"].str.lower() == "debit"
]

vendor_spend = (
    vendor_data
    .groupby("Vendor")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

vendor_transactions = (
    vendor_data
    .groupby("Vendor")
    .size()
)

print("\nTOP VENDORS")
print("-" * 65)

for vendor, amount in vendor_spend.head(5).items():

    transactions = vendor_transactions[vendor]

    print(
        f"{vendor:<22} "
        f"₹{amount:>10,.0f} "
        f"({transactions:,} transactions)"
    )


# ================================================================
# TIME-OF-DAY PATTERNS
# ================================================================

print("\nTIME-OF-DAY PATTERNS")
print("-" * 65)

time_spend = (
    vendor_data
    .groupby("TimePeriod")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

for period, amount in time_spend.items():

    percentage = (
        amount / total_spending * 100
        if total_spending > 0 else 0
    )

    print(
        f"{period:<18} "
        f"₹{amount:>10,.0f} "
        f"({percentage:.1f}%)"
    )

if len(time_spend) > 0:
    print(
        f"\nPeak spending period: "
        f"{time_spend.idxmax()}"
    )


# ================================================================
# TOP CATEGORY MONTHLY TREND
# ================================================================

print("\nMONTHLY TREND")
print("-" * 65)

if len(category_spend) > 0:

    top_category = category_spend.index[0]

    top_category_data = vendor_data[
        vendor_data["category"] == top_category
    ].copy()

    top_category_data["Month"] = (
        top_category_data["Date"]
        .dt.strftime("%b")
    )

    monthly_spend = (
        top_category_data
        .groupby("Month")["Amount"]
        .sum()
    )

    month_order = [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ]

    monthly_spend = monthly_spend.reindex(
        [m for m in month_order if m in monthly_spend.index]
    )

    print(f"{top_category} spending:")

    if len(monthly_spend) > 0:

        max_monthly = monthly_spend.max()

        for month, amount in monthly_spend.items():

            bar_length = (
                int((amount / max_monthly) * 20)
                if max_monthly > 0 else 0
            )

            print(
                f"{month:<5} "
                f"₹{amount:>10,.0f} "
                f"{'#' * bar_length}"
            )


# ================================================================
# TOP ANOMALIES
# ================================================================

print("\nTOP ANOMALIES")
print("-" * 65)

anomaly_count = len(anomalies)

print(f"Total anomalous transactions: {anomaly_count}")

if anomaly_count > 0:

    # Sort by absolute z-score if available
    if "z_score" in anomalies.columns:

        top_anomalies = anomalies.copy()

        top_anomalies["abs_z"] = (
            top_anomalies["z_score"].abs()
        )

        top_anomalies = (
            top_anomalies
            .sort_values("abs_z", ascending=False)
            .head(5)
        )

        for _, row in top_anomalies.iterrows():

            date = row["Date"]
            vendor = row["Vendor"]
            amount = row["Amount"]
            z_score = row["z_score"]

            print(
                f"{date.strftime('%d %b')} - "
                f"{vendor} "
                f"₹{amount:,.0f} "
                f"(z={z_score:.1f})"
            )

    else:

        top_anomalies = (
            anomalies
            .sort_values("Amount", ascending=False)
            .head(5)
        )

        for _, row in top_anomalies.iterrows():

            print(
                f"{row['Date'].strftime('%d %b')} - "
                f"{row['Vendor']} "
                f"₹{row['Amount']:,.0f}"
            )


# ================================================================
# SPENDING ARCHETYPES
# ================================================================

print("\nSPENDING ARCHETYPES")
print("-" * 65)

if archetypes:

    for archetype in archetypes:
        print(f"-> {archetype}")

else:

    print("No strong archetype detected.")


# ================================================================
# KEY INSIGHTS
# ================================================================

print("\nKEY INSIGHTS")
print("-" * 65)


# ------------------------------------------------
# Insight 1 - Spending vs Income
# ------------------------------------------------

if total_income > 0:

    spending_percentage = (
        total_spending / total_income
    ) * 100

    if total_spending > total_income:

        monthly_status = "overspending"

        print(
            f"1. Spending is {spending_percentage:.1f}% "
            f"of total income, indicating {monthly_status}."
        )

    else:

        print(
            f"1. Spending is {spending_percentage:.1f}% "
            f"of total income, leaving room for savings."
        )


# ------------------------------------------------
# Insight 2 - Highest Category
# ------------------------------------------------

if len(category_spend) > 0:

    top_category = category_spend.index[0]
    top_category_amount = category_spend.iloc[0]

    top_category_percentage = (
        top_category_amount / total_spending * 100
        if total_spending > 0 else 0
    )

    print(
        f"2. {top_category} is the highest spending "
        f"category, accounting for "
        f"₹{top_category_amount:,.0f} "
        f"({top_category_percentage:.1f}% of total spending)."
    )

# ------------------------------------------------
# Insight 3 - Top Vendor
# ------------------------------------------------

if len(vendor_spend) > 0:

    top_vendor = vendor_spend.index[0]
    top_vendor_amount = vendor_spend.iloc[0]

    print(
        f"3. {top_vendor} is the highest-spend vendor, "
        f"with ₹{top_vendor_amount:,.0f} spent across "
        f"{vendor_transactions[top_vendor]:,} transactions."
    )

# ------------------------------------------------
# Insight 4 - Peak Month
# ------------------------------------------------

if len(monthly_spend) > 0:

    highest_month = monthly_spend.idxmax()
    highest_month_amount = monthly_spend.max()

    print(
        f"4. {highest_month} recorded the highest "
        f"{top_category} spending at "
        f"₹{highest_month_amount:,.0f}."
    )


# ------------------------------------------------
# Insight 5 - Peak Time
# ------------------------------------------------

if len(time_spend) > 0:

    peak_period = time_spend.idxmax()
    peak_period_amount = time_spend.max()

    peak_percentage = (
        peak_period_amount / total_spending * 100
        if total_spending > 0 else 0
    )

    print(
        f"5. The {peak_period} period has the highest "
        f"spending, accounting for "
        f"{peak_percentage:.1f}% of total spending."
    )


# ================================================================
# FINAL STATUS
# ================================================================

print("\n" + "=" * 65)

if net_change < 0:
    print("              SPENDING > INCOME")
else:
    print("              INCOME > SPENDING")
    print("              ✓ POSITIVE CASH FLOW")

print("=" * 65)

print("\n                    END OF REPORT")
print("=" * 65)
```

## Final Reflection

Building SpendDNA taught me that "messy real-world data" isn't just a warning in a project brief — it's something you only really understand after your code breaks on it.

The hardest part wasn't writing the vendor dictionary or the category mapping — those were mechanical once I inspected the unique descriptions. The real challenge was the date column. My first attempt at `pd.to_datetime()` looked like it worked (no errors thrown), but it silently parsed only 143 out of 1,310 rows correctly. Pandas infers a single date format from the first few rows and quietly fails on everything that doesn't match — even with `errors='coerce'`. I didn't catch this until my monthly summary showed spending in July through December, which was impossible since the dataset only covers January to June. That single bug cascaded through every downstream feature: monthly trends, salary detection, savings rate — all wrong, and none of them threw an error to warn me.

The fix was to parse each of the four known date formats explicitly, one at a time, only applying each format to rows that hadn't already been parsed. It's a good reminder that in data work, the absence of an error message doesn't mean the output is correct — I have to sanity-check the *shape* of my results, not just whether the code ran.

I also learned a smaller but important lesson about consistency: I created near-duplicate columns (`Vendor` and `vendor`, `category` and `Category`) while iterating, which caused several `KeyError`s later. Sticking to one lowercase naming convention from the start, and rebuilding cleanly instead of patching in new cells, would have saved me hours of confusion.

Rahul's numbers confirmed what the brief predicted — he's a Foodie, a Quick Commerce user, an Investor, and a Late-Night Snacker running a deeply negative savings rate. Seeing that pattern emerge from raw, messy transaction strings — not from a clean pre-labeled dataset — is what made this feel like real analytics work rather than a toy exercise.
